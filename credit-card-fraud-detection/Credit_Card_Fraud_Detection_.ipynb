```python
# ============================================================
# CREDIT CARD FRAUD DETECTION & ALERT STRATEGY
# Author: Chirag Gupta
# Dataset: ULB / Worldline Credit Card Fraud Detection
# ============================================================

from pathlib import Path
import json
import hashlib
import zipfile

import numpy as np
import pandas as pd
import joblib

from IPython.display import display
from plotnine import *

from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import HistGradientBoostingClassifier
from sklearn.dummy import DummyClassifier

from sklearn.metrics import (
    average_precision_score,
    precision_recall_curve,
    precision_score,
    recall_score,
    fbeta_score,
    roc_auc_score,
    confusion_matrix,
    classification_report,
    auc
)

from sklearn.inspection import permutation_importance


# ============================================================
# 1. LOAD THE DATASET
# ============================================================

SEED = 42
OUTPUT = Path("/content/fraud_outputs")
OUTPUT.mkdir(exist_ok=True)

csv_path = Path("/content/creditcard.csv")

if not csv_path.exists():
    from google.colab import files

    print("Select creditcard.csv")
    uploaded = files.upload()

    csv_files = [
        name for name in uploaded
        if name.lower().endswith(".csv")
    ]

    if len(csv_files) != 1:
        raise ValueError("Upload exactly one CSV file.")

    csv_path = Path(csv_files[0])

df = pd.read_csv(csv_path)

print("DATASET LOADED")
print("Rows and columns:", df.shape)
display(df.head())


# ============================================================
# 2. DATA QUALITY CHECKS
# ============================================================

expected_columns = (
    ["Time"]
    + [f"V{i}" for i in range(1, 29)]
    + ["Amount", "Class"]
)

if set(df.columns) != set(expected_columns):
    raise ValueError("Dataset columns do not match creditcard.csv.")

df = df[expected_columns].copy()

if not all(
    pd.api.types.is_numeric_dtype(df[column])
    for column in expected_columns
):
    raise ValueError("All dataset columns must be numeric.")

if not np.isfinite(df.to_numpy()).all():
    raise ValueError(
        "Missing or infinite values found. Investigate before training."
    )

if not df["Class"].isin([0, 1]).all():
    raise ValueError("Class must be 0 or 1.")

if (df[["Time", "Amount"]] < 0).any().any():
    raise ValueError("Negative Time or Amount found.")

quality_report = pd.DataFrame({
    "Check": [
        "Total rows",
        "Total columns",
        "Missing values",
        "Exact duplicate rows",
        "Fraud transactions"
    ],
    "Value": [
        len(df),
        df.shape[1],
        int(df.isna().sum().sum()),
        int(df.duplicated().sum()),
        int(df["Class"].sum())
    ]
})

print("\nDATA QUALITY REPORT")
display(quality_report)

quality_report.to_csv(
    OUTPUT / "data_quality_report.csv",
    index=False
)

# Modeling assumption:
# Remove exact full-row duplicates before splitting.
# Without transaction IDs, these cannot be proven to be duplicate events.
# Do not remove large transactions merely because they are outliers.

clean = (
    df.drop_duplicates()
    .sort_values("Time", kind="stable")
    .reset_index(drop=True)
)

clean["Class"] = clean["Class"].astype(int)

print("Rows after duplicate removal:", len(clean))
print("Fraud transactions remaining:", int(clean["Class"].sum()))
print(f"Raw fraud percentage: {100 * df['Class'].mean():.4f}%")
print(f"Clean fraud percentage: {100 * clean['Class'].mean():.4f}%")


# ============================================================
# 3. CHRONOLOGICAL TRAIN / VALIDATION / TEST SPLIT
# ============================================================

# Earlier transactions -> training
# Middle transactions -> validation
# Latest transactions -> final testing
# Equal Time values stay in the same split.

cut_1 = clean["Time"].iloc[int(len(clean) * 0.60)]
cut_2 = clean["Time"].iloc[int(len(clean) * 0.80)]

train = clean.loc[clean["Time"] < cut_1].copy()

validation = clean.loc[
    (clean["Time"] >= cut_1)
    & (clean["Time"] < cut_2)
].copy()

test = clean.loc[clean["Time"] >= cut_2].copy()

for name, part in [
    ("Training", train),
    ("Validation", validation),
    ("Test", test)
]:
    if part["Class"].nunique() != 2:
        raise ValueError(f"{name} must contain fraud and genuine records.")

split_report = pd.DataFrame([
    {
        "Dataset": name,
        "Transactions": len(part),
        "Frauds": int(part["Class"].sum()),
        "Fraud_percentage": 100 * part["Class"].mean()
    }
    for name, part in [
        ("Training", train),
        ("Validation", validation),
        ("Test", test)
    ]
])

print("\nDATA SPLITS")
display(split_report)

split_report.to_csv(OUTPUT / "data_splits.csv", index=False)

features = [
    column for column in expected_columns
    if column != "Class"
]

X_train = train[features]
y_train = train["Class"]

X_validation = validation[features]
y_validation = validation["Class"]

X_test = test[features]
y_test = test["Class"]


# ============================================================
# 4. GGPLOT CHART FUNCTION
# ============================================================

COLORS = {
    "Genuine": "#2563eb",
    "Fraud": "#dc2626"
}

def show_chart(chart, filename):
    chart = (
        chart
        + theme_minimal(base_size=12)
        + theme(figure_size=(9, 5))
    )

    chart.save(
        OUTPUT / f"{filename}.png",
        dpi=160,
        verbose=False
    )

    display(chart)


# Explore training data only.
eda = train.copy()

eda["Transaction_type"] = eda["Class"].map({
    0: "Genuine",
    1: "Fraud"
})

eda["Log_amount"] = np.log1p(eda["Amount"])

# Time is elapsed seconds, not actual clock time.
eda["Elapsed_hour"] = (eda["Time"] // 3600).astype(int)


# ============================================================
# CHART 1: GENUINE VS FRAUD COUNTS
# ============================================================

class_counts = (
    eda.groupby("Transaction_type")
    .size()
    .reset_index(name="Count")
)

show_chart(
    ggplot(
        class_counts,
        aes("Transaction_type", "Count", fill="Transaction_type")
    )
    + geom_col()
    + geom_text(aes(label="Count"), va="bottom")
    + scale_fill_manual(values=COLORS)
    + labs(
        title="Training Data: Genuine vs Fraud Transactions",
        x="Transaction type",
        y="Number of transactions"
    ),
    "01_transaction_counts"
)


# ============================================================
# CHART 2: TRANSACTION AMOUNT DISTRIBUTION
# ============================================================

show_chart(
    ggplot(
        eda,
        aes("Log_amount", fill="Transaction_type")
    )
    + geom_histogram(bins=45, alpha=0.7)
    + facet_wrap("~Transaction_type", scales="free_y")
    + scale_fill_manual(values=COLORS)
    + labs(
        title="Transaction Amount Distribution",
        subtitle="Training data; separate count scales",
        x="log(1 + transaction amount)",
        y="Number of transactions"
    ),
    "02_amount_distribution"
)


# ============================================================
# CHART 3: FRAUD RATE BY ELAPSED HOUR
# ============================================================

hourly = (
    eda.groupby("Elapsed_hour")
    .agg(
        Transactions=("Class", "size"),
        Fraud_transactions=("Class", "sum")
    )
    .reset_index()
)

hourly["Fraud_percentage"] = (
    100 * hourly["Fraud_transactions"] / hourly["Transactions"]
)

show_chart(
    ggplot(hourly, aes("Elapsed_hour", "Fraud_percentage"))
    + geom_line(color="#dc2626")
    + geom_point(color="#dc2626")
    + labs(
        title="Fraud Rate by Elapsed Hour",
        subtitle="Training data; elapsed time is not clock time",
        x="Hours since the first transaction",
        y="Fraud percentage"
    ),
    "03_hourly_fraud_rate"
)


# ============================================================
# CHART 4: FEATURE ASSOCIATIONS WITH FRAUD
# ============================================================

correlations = (
    train[features]
    .corrwith(train["Class"])
    .abs()
    .sort_values(ascending=False)
    .head(10)
)

correlation_data = (
    correlations.rename_axis("Feature")
    .reset_index(name="Absolute_correlation")
)

show_chart(
    ggplot(
        correlation_data,
        aes(
            "reorder(Feature, Absolute_correlation)",
            "Absolute_correlation"
        )
    )
    + geom_col(fill="#7c3aed")
    + coord_flip()
    + labs(
        title="Top Feature Associations with Fraud",
        subtitle="Training data; association does not prove causation",
        x="Feature",
        y="Absolute Pearson correlation"
    ),
    "04_feature_associations"
)


# ============================================================
# 5. TRAIN MODELS
# ============================================================

models = {
    "Always genuine": DummyClassifier(
        strategy="constant",
        constant=0
    ),

    "Logistic regression": make_pipeline(
        StandardScaler(),
        LogisticRegression(
            class_weight="balanced",
            max_iter=2000,
            random_state=SEED
        )
    ),

    "Gradient boosting": HistGradientBoostingClassifier(
        class_weight="balanced",
        max_iter=150,
        max_leaf_nodes=15,
        min_samples_leaf=30,
        l2_regularization=5,
        learning_rate=0.08,
        early_stopping=False,
        random_state=SEED
    )
}

def ranking_metrics(actual, scores):
    precision, recall, _ = precision_recall_curve(actual, scores)

    return {
        "Average_precision": average_precision_score(actual, scores),
        "PR_AUC_trapezoid": auc(recall, precision),
        "ROC_AUC": roc_auc_score(actual, scores)
    }

validation_scores = {}
comparison_rows = []

for name, model in models.items():
    print(f"\nTraining: {name}", flush=True)

    # Scaling and fitting use training data only.
    model.fit(X_train, y_train)

    scores = model.predict_proba(X_validation)[:, 1]
    validation_scores[name] = scores

    comparison_rows.append({
        "Model": name,
        **ranking_metrics(y_validation, scores)
    })

comparison = (
    pd.DataFrame(comparison_rows)
    .sort_values("Average_precision", ascending=False)
)

print("\nVALIDATION MODEL COMPARISON")
display(comparison)

comparison.to_csv(
    OUTPUT / "validation_model_comparison.csv",
    index=False
)

best_name = (
    comparison.loc[comparison["Model"] != "Always genuine"]
    .iloc[0]["Model"]
)

best_model = models[best_name]
validation_score = validation_scores[best_name]

print("Selected using validation average precision:", best_name)


# ============================================================
# CHART 5: MODEL COMPARISON
# ============================================================

show_chart(
    ggplot(
        comparison,
        aes(
            "reorder(Model, Average_precision)",
            "Average_precision"
        )
    )
    + geom_col(fill="#2563eb")
    + coord_flip()
    + labs(
        title="Model Comparison on Validation Data",
        x="Model",
        y="Average precision"
    ),
    "05_model_comparison"
)


# ============================================================
# 6. SELECT ALERT THRESHOLDS ON VALIDATION ONLY
# ============================================================

# Illustrative assumption: review up to 1% of transactions.
# Maximize F2 within this budget: recall receives more weight.
MAX_ALERT_RATE = 0.01
HIGH_PRECISION_TARGET = 0.90

precision, recall, thresholds = precision_recall_curve(
    y_validation,
    validation_score
)

sorted_scores = np.sort(validation_score)

alert_rates = (
    len(sorted_scores)
    - np.searchsorted(sorted_scores, thresholds, side="left")
) / len(sorted_scores)

threshold_report = pd.DataFrame({
    "Threshold": thresholds,
    "Precision": precision[:-1],
    "Recall": recall[:-1],
    "Alert_rate": alert_rates
})

threshold_report["F2"] = (
    5 * threshold_report["Precision"] * threshold_report["Recall"]
    / (
        4 * threshold_report["Precision"]
        + threshold_report["Recall"]
        + 1e-12
    )
)

eligible = threshold_report.loc[
    threshold_report["Alert_rate"] <= MAX_ALERT_RATE
]

if eligible.empty:
    raise ValueError("No feasible threshold within the review budget.")

selected_policy = (
    eligible.sort_values(
        ["F2", "Precision"],
        ascending=False
    )
    .iloc[0]
)

review_threshold = float(selected_policy["Threshold"])

high_candidates = threshold_report.loc[
    (threshold_report["Threshold"] >= review_threshold)
    & (threshold_report["Precision"] >= HIGH_PRECISION_TARGET)
    & (
        threshold_report["Alert_rate"] * len(y_validation)
        >= 5
    )
]

high_threshold = (
    None
    if high_candidates.empty
    else float(
        high_candidates.sort_values("Recall", ascending=False)
        .iloc[0]["Threshold"]
    )
)

print("\nSELECTED VALIDATION POLICY")
display(selected_policy.to_frame("Value"))

print("Review threshold:", review_threshold)
print("High-priority threshold:", high_threshold)

threshold_report.to_csv(
    OUTPUT / "validation_thresholds.csv",
    index=False
)


# ============================================================
# CHART 6: THRESHOLD TRADE-OFFS
# ============================================================

# Thin only the plotted curve, not the threshold calculation.
plot_step = max(1, len(threshold_report) // 1000)

threshold_plot_data = (
    threshold_report.iloc[::plot_step]
    .melt(
        id_vars="Threshold",
        value_vars=["Precision", "Recall", "Alert_rate"],
        var_name="Metric",
        value_name="Value"
    )
)

show_chart(
    ggplot(
        threshold_plot_data,
        aes("Threshold", "Value", color="Metric")
    )
    + geom_line()
    + geom_vline(
        xintercept=review_threshold,
        linetype="dashed"
    )
    + labs(
        title="Validation Threshold Trade-offs",
        subtitle="Dashed line: selected review threshold",
        x="Risk-score threshold",
        y="Rate"
    ),
    "06_threshold_tradeoffs"
)


# ============================================================
# 7. FINAL TEST EVALUATION
# ============================================================

# Model and thresholds are now locked.
# Do not retune them using the test results.

test_score = best_model.predict_proba(X_test)[:, 1]

test_prediction = (
    test_score >= review_threshold
).astype(int)

tn, fp, fn, tp = confusion_matrix(
    y_test,
    test_prediction,
    labels=[0, 1]
).ravel()

test_report = {
    "Model": best_name,
    **ranking_metrics(y_test, test_score),

    "Precision": precision_score(
        y_test, test_prediction, zero_division=0
    ),

    "Recall": recall_score(
        y_test, test_prediction, zero_division=0
    ),

    "F2": fbeta_score(
        y_test, test_prediction, beta=2, zero_division=0
    ),

    "Alert_rate": float(test_prediction.mean()),
    "Genuine_correctly_passed": int(tn),
    "Genuine_falsely_flagged": int(fp),
    "Fraud_missed": int(fn),
    "Fraud_detected": int(tp)
}

print("\nFINAL HELD-OUT TEST RESULTS")
display(pd.DataFrame([test_report]))

print(
    classification_report(
        y_test,
        test_prediction,
        target_names=["Genuine", "Fraud"],
        digits=4,
        zero_division=0
    )
)

pd.DataFrame([test_report]).to_csv(
    OUTPUT / "test_results.csv",
    index=False
)


# ============================================================
# CHART 7: CONFUSION MATRIX
# ============================================================

matrix_data = pd.DataFrame({
    "Actual": ["Genuine", "Genuine", "Fraud", "Fraud"],
    "Predicted": ["Genuine", "Fraud", "Genuine", "Fraud"],
    "Count": [int(tn), int(fp), int(fn), int(tp)]
})

show_chart(
    ggplot(
        matrix_data,
        aes("Predicted", "Actual", fill="Count")
    )
    + geom_tile()
    + geom_text(aes(label="Count"), size=13)
    + scale_fill_gradient(
        low="#eff6ff",
        high="#60a5fa"
    )
    + labs(
        title="Held-out Test Confusion Matrix",
        x="Predicted transaction type",
        y="Actual transaction type"
    ),
    "07_confusion_matrix"
)


# ============================================================
# CHART 8: PRECISION–RECALL CURVE
# ============================================================

test_precision, test_recall, _ = precision_recall_curve(
    y_test,
    test_score
)

pr_data = pd.DataFrame({
    "Recall": test_recall,
    "Precision": test_precision
}).sort_values("Recall")

show_chart(
    ggplot(pr_data, aes("Recall", "Precision"))
    + geom_line(color="#2563eb")
    + geom_hline(
        yintercept=float(y_test.mean()),
        linetype="dashed",
        color="#dc2626"
    )
    + labs(
        title="Held-out Precision–Recall Curve",
        subtitle="Dashed line: fraud prevalence in the test set",
        x="Recall",
        y="Precision"
    ),
    "08_precision_recall"
)


# ============================================================
# 8. FEATURE IMPORTANCE — VALIDATION DATA ONLY
# ============================================================

print("\nCalculating permutation importance...")

importance = permutation_importance(
    best_model,
    X_validation,
    y_validation,
    scoring="average_precision",
    n_repeats=2,
    max_samples=15000,
    random_state=SEED,
    n_jobs=1
)

importance_data = pd.DataFrame({
    "Feature": features,
    "AP_decrease": importance.importances_mean,
    "Standard_deviation": importance.importances_std
}).sort_values("AP_decrease", ascending=False)

display(importance_data.head(10))

importance_data.to_csv(
    OUTPUT / "validation_feature_importance.csv",
    index=False
)


# ============================================================
# CHART 9: FEATURE IMPORTANCE
# ============================================================

show_chart(
    ggplot(
        importance_data.head(10),
        aes("reorder(Feature, AP_decrease)", "AP_decrease")
    )
    + geom_col(fill="#7c3aed")
    + coord_flip()
    + labs(
        title="Feature Reliance on a Validation Sample",
        subtitle="Predictive importance, not a causal explanation",
        x="Feature",
        y="Decrease in average precision"
    ),
    "09_feature_importance"
)


# ============================================================
# 9. FRAUD REVIEW QUEUE
# ============================================================

def score_transactions(transactions):
    missing = set(features) - set(transactions.columns)

    if missing:
        raise ValueError(f"Missing features: {sorted(missing)}")

    inputs = transactions[features].copy()

    if not np.isfinite(inputs.to_numpy()).all():
        raise ValueError("Inputs must contain finite numeric values.")

    scores = best_model.predict_proba(inputs)[:, 1]

    actions = np.where(
        scores >= review_threshold,
        "Review / additional verification",
        "Routine processing"
    )

    if high_threshold is not None:
        actions = np.where(
            scores >= high_threshold,
            "High-priority verification / review",
            actions
        )

    return pd.DataFrame({
        "Risk_score": scores,
        "Suggested_action": actions
    }, index=transactions.index)


# Class is not used as a prediction input.
scored_transactions = score_transactions(X_test)

review_queue = (
    test[["Time", "Amount"]]
    .join(scored_transactions)
    .sort_values("Risk_score", ascending=False)
)

# This is a dataset row index, not a real bank transaction ID.
review_queue.index.name = "Dataset_row_id"

print("\nTOP TRANSACTIONS FOR REVIEW")
display(review_queue.head(20))

review_queue.to_csv(OUTPUT / "transaction_review_queue.csv")


# ============================================================
# 10. POTENTIAL FRAUD EXPOSURE
# ============================================================

actual_fraud = y_test.to_numpy() == 1
flagged = test_prediction == 1

exposure = {
    "Total_test_fraud_amount": float(
        test.loc[actual_fraud, "Amount"].sum()
    ),

    "Flagged_fraud_amount": float(
        test.loc[actual_fraud & flagged, "Amount"].sum()
    ),

    "Missed_fraud_amount": float(
        test.loc[actual_fraud & ~flagged, "Amount"].sum()
    ),

    "Genuine_amount_flagged": float(
        test.loc[~actual_fraud & flagged, "Amount"].sum()
    )
}

print("\nPOTENTIAL EXPOSURE — NOT ACTUAL SAVINGS")
display(pd.DataFrame([exposure]))

pd.DataFrame([exposure]).to_csv(
    OUTPUT / "potential_exposure.csv",
    index=False
)


# ============================================================
# CHART 10: FLAGGED VS MISSED FRAUD AMOUNT
# ============================================================

exposure_chart_data = pd.DataFrame({
    "Outcome": ["Flagged fraud", "Missed fraud"],
    "Amount": [
        exposure["Flagged_fraud_amount"],
        exposure["Missed_fraud_amount"]
    ]
})

show_chart(
    ggplot(
        exposure_chart_data,
        aes("Outcome", "Amount", fill="Outcome")
    )
    + geom_col()
    + labs(
        title="Fraud Transaction Value: Flagged vs Missed",
        subtitle="Potential exposure identified; not actual savings",
        x="Outcome",
        y="Transaction amount"
    ),
    "10_fraud_exposure"
)


# ============================================================
# 11. FRAUD PREVENTION RECOMMENDATIONS
# ============================================================

recommendations = """
PROPOSED FRAUD PREVENTION WORKFLOW

1. Score transactions before authorization using available inputs.

2. Flagged transactions:
   Request additional customer verification or analyst review.

3. High-priority alerts:
   Consider a temporary hold under an approved business policy.
   Do not permanently block a customer solely on this model score.

4. In a real system, add:
   - Card-level transaction frequency and velocity
   - Device and IP changes
   - Merchant risk
   - Authentication outcomes
   These fields are absent from this dataset and were not tested here.

5. Provide a customer verification and appeal route.

6. Use confirmed fraud outcomes for retraining.
   Account for delayed labels.

7. Monitor:
   - Precision and recall
   - Genuine transactions falsely flagged
   - Review workload
   - Missed fraud value
   - Changes in data and model performance

8. Validate on recent, longer-duration data before a live pilot.

LIMITATIONS

- Only two days of historical transactions from 2013.
- V1–V28 are anonymized PCA components.
- No customer, merchant, device or location identifiers.
- Exact-duplicate removal is a modeling assumption.
- Weighted model scores are not calibrated fraud probabilities.
- No real payment blocking or verified financial savings.
- Detection supports prevention; it cannot eliminate all fraud.
"""

print(recommendations)

(OUTPUT / "fraud_prevention_recommendations.txt").write_text(
    recommendations
)


# ============================================================
# 12. SAVE SUMMARY, MODEL AND REPRODUCIBILITY DETAILS
# ============================================================

summary = f"""
CREDIT CARD FRAUD DETECTION — PROJECT SUMMARY

Selected model: {best_name}
Model selection: validation average precision
Evaluation: held-out chronological test set

Average precision: {test_report['Average_precision']:.4f}
Trapezoidal PR-AUC: {test_report['PR_AUC_trapezoid']:.4f}
Precision: {100 * test_report['Precision']:.2f}%
Recall: {100 * test_report['Recall']:.2f}%

Fraud detected: {tp}
Fraud missed: {fn}
Genuine transactions falsely flagged: {fp}
Test transactions flagged: {100 * test_report['Alert_rate']:.2f}%

Flagged fraudulent transaction amount:
{exposure['Flagged_fraud_amount']:.2f}
This represents potential exposure, not actual savings.

Main recommendation:
Use model alerts for additional verification and analyst review.

Dataset source:
https://www.kaggle.com/datasets/mlg-ulb/creditcardfraud

Attribution:
Worldline and ULB Machine Learning Group.
Dal Pozzolo et al. (2015), Calibrating Probability with
Undersampling for Unbalanced Classification, IEEE CIDM.

Important:
Results are historical research findings, not proof of
future or live fraud prevention performance.
"""

print(summary)

(OUTPUT / "project_summary.txt").write_text(summary)

metadata = {
    "random_seed": SEED,
    "dataset_sha256": hashlib.sha256(csv_path.read_bytes()).hexdigest(),
    "selected_model": best_name,
    "features": features,
    "review_threshold": review_threshold,
    "high_priority_threshold": high_threshold,
    "validation_review_budget": MAX_ALERT_RATE,
    "time_cutoffs": [float(cut_1), float(cut_2)],
    "duplicate_policy": "Remove exact full-row duplicates before splitting"
}

(OUTPUT / "run_metadata.json").write_text(
    json.dumps(metadata, indent=2)
)

joblib.dump(
    {
        "model": best_model,
        "metadata": metadata
    },
    OUTPUT / "fraud_detection_model.joblib"
)


# ============================================================
# 13. PACKAGE ALL RESULTS
# ============================================================

zip_path = Path("/content/Fraud_Detection_Results.zip")

with zipfile.ZipFile(
    zip_path,
    "w",
    zipfile.ZIP_DEFLATED
) as archive:

    for file in OUTPUT.iterdir():
        archive.write(file, arcname=file.name)

print("\nPROJECT COMPLETE")
print("Charts, reports and model saved to:", zip_path)
```
